```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import welch
import neurokit2 as nk

# Cargamos los datos del archivo .txt de OpenSignals
datos = np.loadtxt("Basal_1.txt", delimiter=None, comments="#")
eeg = datos[:, 5] # se encuentra en la 5ta columna

# Generate FFT
def fft(signal,fs):
  N = len(signal)
  T = 1/fs
  signal_fft = np.fft.rfft(signal)
  signal_fft_mag = np.abs(signal_fft)
  signal_fft_dB = 20*np.log10(signal_fft_mag)
  freqs = np.fft.rfftfreq(N, T)
  return freqs,signal_fft,signal_fft_mag, signal_fft_dB

#Datos
fs = 100
T = 1/fs

# Segmento 0–20 s
fin = int(20*fs)
eeg_segment = eeg[0:fin]

N = len(eeg_segment)
t = np.arange(N) / fs

eeg_segment = eeg_segment - np.mean(eeg_segment)

#FFT
f_hz,eeg_fft,eeg_fft_mag,eeg_fft_dB = fft(eeg_segment,fs)

#PSD
f_eeg, psd_eeg = welch(eeg, fs=fs, nperseg=fs*2)

plt.figure(figsize=(12, 10))
plt.subplot(4,1,1)
plt.plot(t,eeg_segment)
plt.title("EEG - Basal 1")
plt.grid()
plt.xlabel("Tiempo (s)")
plt.ylabel("Amplitud (uV)")

plt.subplot(4,1,2)
plt.plot(f_hz,eeg_fft_mag)
plt.title("FFT - Basal 1 ")
plt.grid()
plt.xlabel("Frequencia (Hz)")
plt.ylabel("Magnitud")

plt.subplot(4,1,3)
plt.plot(f_hz,eeg_fft_dB)
plt.title("FFT dB - Basal 1 ")
plt.xlabel("Frequencia (Hz)")
plt.ylabel("Magnitud [dB]")
plt.grid()

plt.subplot(4,1,4)
plt.plot(f_eeg,psd_eeg)
plt.title("PSD - Basal 1 ")
plt.xlabel("Frequencia (Hz)")
plt.ylabel("Magnitud [dB]")
plt.grid()

plt.tight_layout()
plt.show()

# Análisis por bandas
bandas = {"Delta (0.5–4 Hz)": (0.5, 4),"Theta (4–8 Hz)": (4, 8),"Alfa (8–12 Hz)": (8, 12),"Beta (13–30 Hz)": (13, 30),"Gamma (30–70 Hz)": (30, 40)}

potencias = []
for min, max in bandas.values():
    idx = np.logical_and(f_eeg >= min, f_eeg <= max)
    potencia = np.trapezoid(psd_eeg[idx], f_eeg[idx])
    potencias.append(potencia)

colores = ["#AEC6CF", "#FFB347", "#B39EB5", "#77DD77", "#FF6961"]

plt.figure(figsize=(8,4))
plt.bar(bandas.keys(), potencias, color=colores)
plt.title("Potencia por banda - EEG basal")
plt.ylabel("Potencia")
plt.xticks(rotation=45)
plt.grid(axis="y")
plt.ylim(0, 1000)
plt.show()

# Limpiar señal EEG
clean = nk.signal_filter(
    eeg_segment,
    sampling_rate=fs,
    lowcut=1,
    highcut=40,
    method="butterworth",
    order=5
)

# Graficar original y filtrada
plt.figure(figsize=(12,4))

plt.plot(eeg_segment, alpha=0.5, label="Original")
plt.plot(clean, label="Filtrada")
plt.grid()
plt.legend()
plt.title("EEG")
plt.show()
```
